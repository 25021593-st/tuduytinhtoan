```python
so_thuc = float(input("Nhập vào một số thực: "))
tron_len = lam_tron_len(so_thuc)
tron_xuong = lam_tron_xuong(so_thuc)
tron_gan_nhat = lam_tron_gan_nhat(so_thuc)
print(tron_len, tron_xuong, tron_gan_nhat)
def lam_tron_xuong(n):
    if n >= 0:
        return int(n)
    else:
        return int(n) - 1 if n != int(n) else int(n)

def lam_tron_len(n):
    if n > 0:
        return int(n) + 1 if n != int(n) else int(n)
    else:
        return int(n)

def lam_tron_gan_nhat(n):
    phan_thap_phan = abs(n - int(n))
    if phan_thap_phan < 0.5:
        return int(n)
    elif phan_thap_phan > 0.5:
        return int(n) + 1 if n > 0 else int(n) - 1
    else:
        nguyen = int(n)
        return nguyen if nguyen % 2 == 0 else (nguyen + 1 if n > 0 else nguyen - 1)
```

```python
a = int(input("Nhập số a: "))
b = int(input("Nhập số b: "))
c = int(input("Nhập số c: "))
d = int(input("Nhập số d: "))
def cap_so_nhan(a, b, c, d):
    if a == 0:
        return "NO" 
    if b % a != 0:
        return "NO"
    cong_boi = b // a
    if c == b * cong_boi and d == c * cong_boi:
        return "YES"
    else:
        return "NO"
ket_qua = cap_so_nhan(a, b, c, d)
print(ket_qua)
```
