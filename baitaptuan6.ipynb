```python
nums = list(map(int, input().split())) #nhập số nguyên dương cách nhau bởi dấu cách
result = [] #danh sách lưu trữ các số không trùng lặp
for x in nums: #duyệt qua từng số trong danh sách ban đầu
    if x not in result: #nếu số chưa có trong danh sách kết quả
        result.append(x) #thêm số vào danh sách kết quả
print(result) #in danh sách các số không trùng lặp
```

```python
nums = list(map(int, input().split())) #nhập số nguyên dương cách nhau bởi dấu cách
cum_sum = [] #danh sách lưu trữ tổng tích lũy
current_total = 0 #biến lưu tổng hiện tại
for x in nums: #duyệt qua từng số trong danh sách ban đầu
    current_total += x #cập nhật tổng hiện tại
    cum_sum.append(current_total) #thêm tổng hiện tại vào danh sách tổng tích lũy
print(cum_sum) #in danh sách tổng tích lũy
```

```python
tup = tuple(map(int, input().split())) #nhập tuple số nguyên dương cách nhau bởi dấu cách   
k = int(input()) 
n = len(tup)
if n > 0:
    k = k % n
    result = tup[k:] + tup[:k] 
    print(result)
else:
    print(tup)
```

```python
pairs = input().split()
d = {}
for p in pairs:
    key, value = p.split(':')
    if key not in d:
        d[key] = [value]
    else:
        d[key].append(value)
print(d)
```

```python
nums = list(map(int, input().split()))
res = {'positives': 0, 'negatives': 0, 'zeros': 0}
for x in nums:
    if x > 0:
        res['positives'] += 1
    elif x < 0:
        res['negatives'] += 1
    else:
        res['zeros'] += 1
print(res)
```

```python

dict1_input = input("dict 1 (k:v): ").split()
dict2_input = input("dict 2 (k:v): ").split()

def parse_dict(pairs):
    d = {}
    for p in pairs:
        k, v = p.split(':')
        d[k] = int(v)
    return d

d1 = parse_dict(dict1_input)
d2 = parse_dict(dict2_input)

# Gộp và cộng giá trị
combined = d1.copy()
for k, v in d2.items():
    combined[k] = combined.get(k, 0) + v

# In theo thứ tự từ điển của key
sorted_dict = {k: combined[k] for k in sorted(combined)}
print(sorted_dict)
```
