```python
# Bài W4A6: Tính tổng các số nguyên tố trong đoạn [a, b]

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
a, b = map(int, input(" a và b (a ≤ b): ").split())

if a > b or a < 1 or b < 1:
    print("không hợp lệ")
else:
    tong = 0
    for i in range(a, b + 1):
        if is_prime(i):
            tong += i
    print(f"Tổng các số nguyên tố trong đoạn [{a}, {b}] là: {tong}")
```

```python
def is_prime(x):
    if x < 2:
        return False
    for i in range(2, int(x**0.5) + 1):
        if x % i == 0:
            return False
    return True

n = int(input("Nhập số nguyên dương n (n ≥ 2): "))
i = 2
largest = 0

while i <= n:
    if n % i == 0 and is_prime(i):
        largest = i
        n //= i
    else:
        i += 1

print("Ước số nguyên tố lớn nhất là:", largest)
```

```python
def is_palindrome(x):
    return str(x) == str(x)[::-1]

def reverse_number(x):
    return int(str(x)[::-1])

n = int(input("Nhập số nguyên dương n: "))
steps = 0

while not is_palindrome(n):
    n += reverse_number(n)
    steps += 1

print(f"Số bước: {steps}")
print(f"Giá trị palindrome thu được: {n}")
```

```python
def has_distinct_digits(x):
    digits = str(x)
    return len(set(digits)) == len(digits)

n = int(input("Nhập số nguyên dương n: "))
found = False

for i in range(1, int(n**0.5) + 1):
    square = i * i
    if square <= n and has_distinct_digits(square):
        print(square, end=" ")
        found = True

if not found:
    print("no number")
```

```python
def collatz_length(x):
    count = 1
    while x != 1:
        x = x // 2 if x % 2 == 0 else 3 * x + 1
        count += 1
    return count

n = int(input("Nhập số nguyên dương n: "))
max_len = 0
best_x = 1

for i in range(1, n + 1):
    length = collatz_length(i)
    if length > max_len or (length == max_len and i < best_x):
        max_len = length
        best_x = i

print(f"{best_x} {max_len}")
```
